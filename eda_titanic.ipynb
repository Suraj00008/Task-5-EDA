```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df["Survived"].value_counts()
```

```python
df["Age"].hist(figsize=(8,5))
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
sns.boxplot(x=df["Fare"])
plt.title("Fare Distribution")
plt.show()
```

```python
sns.countplot(x="Sex", hue="Survived", data=df)
plt.title("Survival by Gender")
plt.show()
```

```python
sns.countplot(x="Pclass", hue="Survived", data=df)
plt.title("Survival by Passenger Class")
plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(df[["Survived","Age","Fare","Pclass"]])
plt.show()
```

```python
sns.scatterplot(
    x="Age",
    y="Fare",
    hue="Survived",
    data=df
)

plt.title("Age vs Fare")
plt.show()
```

# Summary of Findings

1. Most passengers were between 20 and 40 years old.
2. Fare contains several outliers.
3. Female passengers survived more often than males.
4. First-class passengers had the highest survival rate.
5. Higher ticket fares were associated with better survival chances.
6. Missing values were present in Age, Cabin, and Embarked columns.
7. Fare and Passenger Class showed a significant relationship.
